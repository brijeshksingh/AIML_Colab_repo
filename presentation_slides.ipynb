## **Slide 1: Project Overview**
**Objective:** To build and evaluate a deep learning model for classifying Twitter sentiment into four categories: Positive, Negative, Neutral, and Irrelevant.

**Key Technologies Used:**
- **Programming Language:** Python
- **Core Libraries:** TensorFlow, Keras, NLTK, Scikit-learn, Pandas
- **Model Architecture:** Bidirectional Long Short-Term Memory (LSTM)
- **Environment:** Jupyter Notebook

**End-to-End Pipeline:**
1.  **Data Loading & Cleaning:** Handled missing values, duplicates, and irrelevant text.
2.  **Preprocessing:** Applied tokenization, stop-word removal, and lemmatization.
3.  **EDA:** Visualized data patterns to derive insights.
4.  **Modeling:** Built and configured a Bidirectional LSTM model.
5.  **Evaluation:** Established a framework to measure performance.

## **Slide 3: Data Preprocessing Pipeline**

**Objective:** Transform raw tweet text into a clean, numerical format suitable for the RNN model.

1.  **Text Cleaning:**
    - Removed URLs, user mentions (@), and hashtags (#).
    - Stripped punctuation and numerical digits.
    - Converted all text to lowercase.

2.  **Text Normalization (NLTK):**
    - **Tokenization:** Split text into individual words (tokens).
    - **Stop-Word Removal:** Eliminated common words with little semantic value (e.g., 'the', 'a', 'is').
    - **Lemmatization:** Reduced words to their base or dictionary form (e.g., 'running' -> 'run').

3.  **Numerical Conversion (Keras):**
    - **Tokenization:** Mapped each unique word to an integer.
    - **Padding:** Ensured all text sequences have a uniform length of **100 tokens**.

## **Slide 6: Model Architecture - Bidirectional LSTM**

**Why Bidirectional LSTM?** It processes text from left-to-right and right-to-left, allowing it to capture context from both past and future words in a sequence. This is crucial for understanding nuanced sentiment.

#### **Layer-by-Layer Breakdown:**
1.  **Embedding Layer:** Converts integer-based vocabulary into dense vectors of **128 dimensions**.
2.  **Bidirectional LSTM Layer:** The core processing unit with **128 LSTM cells** per direction.
3.  **Global Max Pooling:** Extracts the most significant features from the LSTM output.
4.  **Dense Layers:** Two fully connected layers (128 and 64 neurons) with ReLU activation for final classification.
5.  **Output Layer:** A `softmax` layer that outputs probabilities for the 4 sentiment classes.

| :--- | :--- |
| **Trainable Parameters** | 2,225,476 |
| **Regularization** | Dropout (30%) & Batch Normalization |

## **Slide 7: Model Configuration & Training Setup**

#### **Training Configuration:**
- **Optimizer:** Adam (Learning Rate: 0.001)
- **Loss Function:** Categorical Crossentropy (suitable for multi-class classification).
- **Batch Size:** 64
- **Epochs:** 10 (maximum)

#### **Advanced Training Techniques:**
- **Early Stopping:** Training automatically stops if validation accuracy does not improve for 5 consecutive epochs, preventing overfitting.
- **Learning Rate Scheduling:** The learning rate is reduced if the validation loss plateaus, allowing for finer adjustments.
- **Class Weights:** Applied to give more importance to under-represented classes during training, counteracting the class imbalance.

## **Slide 8: Evaluation Framework**

**Objective:** To rigorously assess the model's performance on unseen data.

#### **Data Split:**
- **Training Set:** 70%
- **Validation Set:** 15%
- **Testing Set:** 15%
*A stratified split was used to ensure each set had a similar sentiment distribution.*

#### **Primary Performance Metrics:**
- **Accuracy:** The overall percentage of correct predictions.
- **Precision:** Of all the tweets predicted as positive, how many were actually positive?
- **Recall:** Of all the actual positive tweets, how many did the model correctly identify?
- **F1-Score:** The harmonic mean of Precision and Recall, providing a single score for model quality.
- **Confusion Matrix:** A table visualizing correct vs. incorrect predictions for each class.
